```python
def load_fasta(filepath):

    sequences = {}

    with open(filepath) as f:

        seq_id = None
        seq_lines = []

        for line in f:
            line = line.strip()

            if line.startswith(">"):

                if seq_id:
                    sequences[seq_id] = "".join(seq_lines)

                seq_id = line[1:]
                seq_lines = []

            else:
                seq_lines.append(line)

        if seq_id:
            sequences[seq_id] = "".join(seq_lines)

    return sequences
```

```python
def align_sequences(seq1, seq2):

    length = min(len(seq1), len(seq2))

    matches = 0

    for i in range(length):
        if seq1[i] == seq2[i]:
            matches += 1

    percent_identity = matches / length * 100

    return matches, percent_identity
```

```python
def find_best_match(query, database):

    best_id = None
    best_score = -1
    best_identity = 0

    for seq_id, sequence in database.items():

        score, identity = align_sequences(query, sequence)

        if score > best_score:
            best_score = score
            best_identity = identity
            best_id = seq_id

    return best_id, best_score, best_identity
```

```python
def main():

    args = parse_arguments()

    database = load_fasta(args.db)

    query_sequences = load_fasta(args.query)
    query = list(query_sequences.values())[0]

    best_id, score, identity = find_best_match(query, database)

    print("Best match:", best_id)
    print("Alignment score:", score)
    print("Percent identity:", f"{identity:.2f}%")
```

```python
#Validate Query sequence
#Check only contains A,T, C, G, not empty. If invalid -> print error and exit
def validate_sequence(sequence: str) -> str:
    """
    Validate that a DNA sequence contains only A, T, C and G.

    Parameters:
        sequence (str): The DNA sequence provided.
    Returns:
        str: The validated sequence converted to uppercase.
    Raises:
        ValueError if the sequence is empty or contains invalid characters.
        
    """
    if not sequence:
        raise ValueError("Query sequence is empty.")
    
    sequence = sequence.upper()
    valid_bases = {"A", "T", "C", "G"}

    invalid_bases = set(sequence) - valid_bases
    if invalid_bases:
        raise ValueError(f"Invalid Character(s) found in sequence: {', '.join(invalid_bases)}")
        
    return sequence    
```
